## Operating Mask Creation

In this step, an operating mask (`Op_Mask`) is created to distinguish periods where the photovoltaic system is expected to actively produce power.

Instead of using all pyranometers, only the two Plane-of-Array (POA) sensors are used:

- `Pyra1_Wm2_Avg` → West-tilted POA irradiance
- `Pyra3_Wm2_Avg` → East-tilted POA irradiance

The Global Horizontal Irradiance sensor (`Pyra2_Wm2_Avg`) is excluded because it does not directly represent irradiance incident on the tilted PV modules.

The maximum value between the two POA sensors is used:

```python
poa_max = canopy_2017[['Pyra1_Wm2_Avg', 'Pyra3_Wm2_Avg']].max(axis=1)
```

This accounts for the east-west panel orientation, ensuring that if either side of the system receives sufficient irradiance, the system is considered operational.

An operating mask is then generated using a threshold of:

```text
100 W/m²
```

- `1` → System operational
- `0` → Nighttime or low-irradiance conditions

This prevents naturally low nighttime production from being incorrectly interpreted as system faults.

````python
poa_max = canopy_2017[['Pyra1_Wm2_Avg', 'Pyra3_Wm2_Avg']].max(axis=1)

canopy_2017['Op_Mask'] = np.where(poa_max > 100, 1, 0)
````

# Expected vs Actual Power Analysis (Performance Ratio)

In this step, physics-based features are engineered to compare the photovoltaic system’s expected power production against its actual inverter output.

The expected power is estimated using:

- Plane-of-array irradiance from reference cell measurements
- Module temperature
- PV system rated DC capacity
- Temperature correction coefficient
- Inverter AC power ceiling

The theoretical DC power is calculated using the IEC 61724-1 temperature-corrected PV performance model:

$$
P_{\text{expected,DC}} =
P_{\text{rated,DC}}
\times
\left(
\frac{\text{POA}_{\text{fused}}}{1000}
\right)
\times
\left[
1 + \gamma (T_{\text{module}} - T_{\text{STC}})
\right]
$$

The expected DC power is then clipped to the inverter AC rating in order to produce a physically realistic AC-side expected power estimate.

Reference cell measurements are fused using the maximum value between the two sensors to account for the east-west PV orientation and reduce the impact of temporary sensor disagreement.

Additional validity constraints are applied using IEC 61724-1 recommendations:

- PR calculations are only considered reliable above:
  ```text
  200 W/m²
  ```
- The inverter must also be in a valid operational state

This produces the `PR_evaluable` mask used during fault labeling.

---

### Engineered Features

Several physics-based diagnostic features are generated:

| Feature | Description |
|---|---|
| `P_expected_DC_kW` | Temperature-corrected expected DC power |
| `P_expected_AC_kW` | AC-bounded expected power |
| `PR` | Instantaneous Performance Ratio |
| `PR_roll_15min_weighted` | Energy-weighted rolling PR |
| `PR_roll_std_15min` | Rolling PR volatility |
| `power_deficit_kW` | Difference between expected and actual power |

Energy-weighted rolling PR is used instead of a simple rolling mean because it reduces instability during low-irradiance periods and better matches the IEC 61724-1 definition of period-level performance ratio.

The resulting features provide physically meaningful indicators for:

- PV underperformance detection
- Fault analysis
- System degradation monitoring
- Time-series anomaly detection
- Machine learning model training

````python
canopy_2017 = pd.read_csv('/content/drive/MyDrive/NIST_processed/canopy_2017_preprocessing.csv')
````

````python
# Convert the TIMESTAMP column to pandas datetime object
canopy_2017['TIMESTAMP'] = pd.to_datetime(canopy_2017['TIMESTAMP'])

# Set TIMESTAMP as index
canopy_2017.set_index('TIMESTAMP', inplace=True)
````

````python
P_RATED_DC_KW  = 243.0    # DC nameplate capacity
P_AC_MAX_KW    = 260.0    # Inverter AC rated output
GAMMA          = -0.00485 # Power temperature coefficient for monocrystalline Si (/°C)
T_STC          = 25.0     # Standard Test Condition reference temperature (°C)

VOC_STC           = 37.0  # open circuit voltage at STC (V)
VMP_STC           = 30.0  # maximum power voltage at STC (V)
ISC_STC           = 8.60  # short circuit current at STC (A)
IMP_STC           = 7.84  # maximum power current at STC (A)
NOCT              = 47.5  # nominal operating cell temperature (°C)
MODULES_PER_STRING = 12
N_CELLS_PER_MODULE = 60

# IEC 61724-1 minimum irradiance for valid PR calculations
# Below this threshold, measurement noise dominates relative errors
# This is seperate from Op_Mask (100 W/m²) which defines operational hours
POA_MIN_PR     = 200.0    # W/m² — IEC 61724-1


# 1. POA Irradiance
# Using only the two reference cells (spectral response matches modules)
# rather than mixing with flat-spectral pyranometers.
# Mean is robust to the tilt of the array.
canopy_2017['POA_fused'] = canopy_2017[['RefCell1_Wm2_Avg', 'RefCell2_Wm2_Avg']].mean(axis=1)

# IEC 61724-1 validity flag — PR calculations only trustworthy above 200 W/m²
canopy_2017['PR_evaluable'] = ((canopy_2017['Op_Mask'] == 1) & (canopy_2017['POA_fused'] >= POA_MIN_PR) &
                               (canopy_2017['InvOpStatus_Avg'] > 0) & (canopy_2017['InvOpStatus_Avg'] <= 5)).astype(int)


# 2. Temperature-Corrected Expected Power
# Standard IEC 61724-1:2021:
# P_expected = P_rated × (poa_fused / 1000) × [1 + γ × (T_mod - T_STC)]
# This is the power the system should be producing given current irradiance and module temperature.
# Shortfalls indicate a fault signal.

# Include only normal functioning RTDs
rtd_cols = ['RTD_C_Avg_1', 'RTD_C_Avg_2', 'RTD_C_Avg_4', 'RTD_C_Avg_8', 'RTD_C_Avg_9']
canopy_2017['module_temp'] = canopy_2017[rtd_cols].median(axis=1)

canopy_2017['P_expected_DC_kW'] = (P_RATED_DC_KW * (canopy_2017['POA_fused'] / 1000.0) * (1 + GAMMA * (canopy_2017['module_temp'] - T_STC)))


# 3. AC-Bounded Expected Power
# The numerator of PR is AC output, but the raw P_expected_DC can exceed the inverter's AC ceiling on high-irradiance days.

canopy_2017['P_expected_AC_kW'] = canopy_2017['P_expected_DC_kW'].clip(upper=P_AC_MAX_KW)


# 4. Performance Ratio
# PR = P_actual_AC / P_expected
#
# Threshold justification (all based on Op_Mask=1 AND POA ≥ 200 W/m²):
#
#   > 1.05    : Suspiciously high - but possible
#
#   0.85–1.05 : NORMAL — within expected real-world loss range.
#
#   0.70–0.85 : UNDERPERFORMANCE — measurable degradation beyond normal losses.
#
#   0.50–0.70 : SEVERE UNDERPERFORMANCE — likely 1–2 string outages, significant soiling event, or partial inverter fault.
#               1 string out of 7 → ~14% loss → PR drops from 0.82 to ~0.71.
#               2 strings → ~28% loss → PR ≈ 0.59. Threshold placement is
#
#   0.00–0.50 : CRITICAL UNDERPERFORMANCE — near-total production loss during operational hours.
#               Consistent with inverter shutdown or 3+ strings out.
#
#   < 0       : NEGATIVE AC POWER — inverter drawing from grid (standby/housekeeping).
#               Normal at night; abnormal during Op_Mask=1 hours.

canopy_2017['PR'] = canopy_2017['InvPAC_kW_Avg'] / canopy_2017['P_expected_AC_kW'].replace(0, np.nan)


# 5. Energy-Weighted Rolling PR
# Plain rolling mean of PR gives equal weight to low- and high-irradiance minutes.
# At low irradiance, small absolute errors produce large relative PR noise.
# The energy-weighted formulation matches IEC 61724-1 period-level PR definition:
#   PR_weighted = Σ(P_actual) / Σ(P_expected)  over the window
#
# This naturally down-weights low-production minutes because they contribute less to both numerator and denominator in absolute terms.
# Single consistent window (15 min) used for all thresholds — eliminates the inconsistency of using different windows for different severity levels.

p_actual_masked   = canopy_2017['InvPAC_kW_Avg'].where(canopy_2017['PR_evaluable'] == 1)
p_expected_masked = canopy_2017['P_expected_AC_kW'].where(canopy_2017['PR_evaluable'] == 1)

canopy_2017['PR_roll_15min_weighted'] = p_actual_masked.rolling(window=15, min_periods=8).sum() / p_expected_masked.rolling(window=15, min_periods=8).sum()

# Additional windows for feature richness (used as ML features, not for labels)
canopy_2017['PR_roll_5min_weighted'] = p_actual_masked.rolling(window=5, min_periods=3).sum() / p_expected_masked.rolling(window=5, min_periods=3).sum()

canopy_2017['PR_roll_60min_weighted'] = p_actual_masked.rolling(window=60, min_periods=30).sum() / p_expected_masked.rolling(window=60, min_periods=30).sum()

# Rolling volatility of PR — spiky PR = intermittent fault; sustained low = chronic fault
canopy_2017['PR_roll_std_15min'] = canopy_2017['PR'].rolling(window=15, min_periods=8).std()

# Power deficit — absolute kW gap, useful for severity ranking in labeling merge
canopy_2017['power_deficit_kW'] = canopy_2017['P_expected_AC_kW'] - canopy_2017['InvPAC_kW_Avg']
````

````python
round_cols = [
    'module_temp', 'P_expected_DC_kW', 'P_expected_AC_kW', 'PR',
    'PR_roll_5min_weighted', 'PR_roll_15min_weighted', 'PR_roll_60min_weighted',
    'PR_roll_std_15min', 'power_deficit_kW'
]

canopy_2017[round_cols] = canopy_2017[round_cols].round(3)
````

## Physics-Based Fault Labeling

In this step, rule-based fault labels are generated using the engineered Performance Ratio (PR) features.

The labeling process compares the photovoltaic system’s actual AC output against the theoretically expected power under current irradiance and temperature conditions.

To reduce short-term noise and transient fluctuations, labeling is based on the:

```python
PR_roll_15min_weighted
```

feature rather than instantaneous PR values.

Only rows satisfying the IEC 61724-1 validity conditions (`PR_evaluable = 1`) are considered suitable for reliable PR evaluation.

---

### Label Definitions

| Label | Description |
|---|---|
| `NOT_EVALUABLE` | Nighttime, low irradiance, or invalid inverter operating state |
| `MISSING_DATA` | Missing critical measurements required for PR calculation |
| `CRITICAL_UNDERPERFORMANCE` | Near-total production loss (`0.00 ≤ PR < 0.50`) |
| `SEVERE_UNDERPERFORMANCE` | Significant sustained underperformance (`0.50 ≤ PR < 0.70`) |
| `UNDERPERFORMANCE` | Moderate underperformance (`0.70 ≤ PR < 0.85`) |
| `SUSPICIOUSLY_HIGH` | PR values above physically expected range (`PR > 1.05`) |
| `NEGATIVE_AC_POWER` | Negative inverter AC power during operational hours |
| `NORMAL` | System operating within expected performance range |

---

### Threshold Rationale

The selected PR thresholds are based on expected real-world photovoltaic system behavior:

- Healthy crystalline silicon PV systems typically operate around:
  ```text
  PR ≈ 0.80–0.90
  ```

- Moderate PR reductions may indicate:
  - Soiling
  - Partial shading
  - Degradation
  - Minor string faults

- Severe PR reductions are consistent with:
  - String outages
  - Inverter faults
  - Major production losses

Negative AC power during daylight conditions is treated separately because it may indicate abnormal inverter standby or grid-draw behavior.

---

### Purpose

These labels act as weak physics-based supervision signals that can later be used for:

- Fault detection experiments
- Supervised machine learning
- Anomaly detection validation
- PV system performance monitoring
- Comparative evaluation against inverter fault codes

````python
#Fault Labeling

# Priority order: NOT_EVALUABLE > INVERTER_SHUTDOWN > MISSING_DATA > CRITICAL_UNDERPERFORMANCE > SEVERE_UNDERPERFORMANCE >
#                 UNDERPERFORMANCE > SUSPICIOUSLY_HIGH > NEGATIVE_AC_POWER > NORMAL

pr = canopy_2017['PR_roll_15min_weighted']
evaluable = canopy_2017['PR_evaluable'] == 1

conditions = [
    # 1. Outside operating window or below irradiance floor
    canopy_2017['PR_evaluable'] == 0,

    # 2. Cannot compute PR — critical input is missing
    canopy_2017['P_expected_AC_kW'].isna() | canopy_2017['InvPAC_kW_Avg'].isna() | (evaluable & canopy_2017['PR_roll_15min_weighted'].isna()),

    # 3. Critical underperformance: PR 0.00–0.50
    # Near-total production loss — inverter shutdown or 3+ strings out
    evaluable & (pr >= 0.0) & (pr < 0.50),

    # 4. Severe underperformance: PR 0.50–0.70
    # Consistent with 1–2 string outages or significant partial shading
    evaluable & (pr >= 0.50) & (pr < 0.70),

    # 5. Underperformance: PR 0.70–0.85
    # Below realistic expectation for healthy c-Si system
    evaluable & (pr >= 0.70) & (pr < 0.85),

    # 6. Suspiciously high: PR > 1.05
    # Physically possible but rare
    evaluable & (pr > 1.05),

    # 7. Negative AC power during operational hours
    # Inverter drawing from grid — abnormal during daylight
    evaluable & (canopy_2017['InvPAC_kW_Avg'] < 0)
]

choices = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'CRITICAL_UNDERPERFORMANCE',
    'SEVERE_UNDERPERFORMANCE',
    'UNDERPERFORMANCE',
    'SUSPICIOUSLY_HIGH',
    'NEGATIVE_AC_POWER'
]

canopy_2017['label_power'] = np.select(conditions, choices, default='NORMAL')


# Check results
print("=== Labeling: Power Performance ===\n")

print("PR distribution (PR_evaluable = 1 rows only):")
print(canopy_2017.loc[evaluable, 'PR'].describe().round(4))

print(f"\nWeighted rolling PR distribution (PR_evaluable = 1):")
print(canopy_2017.loc[evaluable, 'PR_roll_15min_weighted'].describe().round(4))

print(f"\nlabel_power distribution:")
print(canopy_2017['label_power'].value_counts())
print(f"\nAs percentage of total rows:")
print((canopy_2017['label_power'].value_counts() / len(canopy_2017) * 100).round(2))

print(f"\nPR_evaluable breakdown:")
print(f"  Op_Mask=1 rows total          : {(canopy_2017['Op_Mask']==1).sum()}")
print(f"  PR_evaluable=1 (POA ≥ 200)   : {(canopy_2017['PR_evaluable']==1).sum()}")
print(f"  Rows lost to IEC floor        : {((canopy_2017['Op_Mask']==1) & (canopy_2017['PR_evaluable']==0)).sum()}")

print(f"\nNEGATIVE_AC_POWER during PR_evaluable hours:")
neg = (canopy_2017['label_power'] == 'NEGATIVE_AC_POWER')
print(f"  Count: {neg.sum()} rows ({neg.sum()/60:.1f} hours)")
````

#String Level Faults

````python
canopy_2017 = pd.read_csv('/content/drive/MyDrive/NIST_processed/canopy_2017_Label.csv')
````

````python
# Convert the TIMESTAMP column to pandas datetime object
canopy_2017['TIMESTAMP'] = pd.to_datetime(canopy_2017['TIMESTAMP'])

# Set TIMESTAMP as index
canopy_2017.set_index('TIMESTAMP', inplace=True)
````

````python
string_current_cols = [f'ShuntCurrent_A_Avg_{i}' for i in range(1, 8)]
string_power_cols   = [f'ShuntPDC_kW_Avg_{i}'    for i in range(1, 8)]

# Apply physical bounds to string currents
# PV string current cannot be negative or exceed module Isc * parallel strings
CURRENT_MIN = 0.0    # A — negative current is physically impossible for DC strings
CURRENT_MAX = 150.0  # A — ceiling above observed max of ~139A

for col in string_current_cols:
    canopy_2017[col] = canopy_2017[col].where(
        canopy_2017[col].between(CURRENT_MIN, CURRENT_MAX), other=np.nan
    )

# Same for string power
POWER_MIN = 0.0
POWER_MAX = 50.0    # kW per string — 243kW / 7 strings ≈ 34.7 kW max
for col in string_power_cols:
    canopy_2017[col] = canopy_2017[col].where(
        canopy_2017[col].between(POWER_MIN, POWER_MAX), other=np.nan
    )
````

````python
VOC_STC           = 37.0  # open circuit voltage at STC (V)
VMP_STC           = 30.0  # maximum power voltage at STC (V)
ISC_STC           = 8.60  # short circuit current at STC (A)
IMP_STC           = 7.84  # maximum power current at STC (A)
NOCT              = 47.5  # nominal operating cell temperature (°C)
N_STRINGS         = 7
STRING_CV_THRESHOLD = 0.15 # coefficient of variation threshold
MODULES_PER_STRING  = 12
N_CELLS_PER_MODULE  = 60
POA_MIN_STRING      = 200 # W/m² — minimum irradiance to evaluate string faults
OPEN_CIRCUIT_RATIO  = 0.10 # string is "dead" if < 10% of mean


# 1. String Current Aggregates
canopy_2017['I_string_mean']  = canopy_2017[string_current_cols].mean(axis=1)
canopy_2017['I_string_std']   = canopy_2017[string_current_cols].std(axis=1)
canopy_2017['I_string_min']   = canopy_2017[string_current_cols].min(axis=1)
canopy_2017['I_string_max']   = canopy_2017[string_current_cols].max(axis=1)

# Coefficient of variation — spread relative to mean, scale-independent
# High CV = strings behaving differently from each other
canopy_2017['I_string_cv'] = canopy_2017['I_string_std'] / canopy_2017['I_string_mean'].replace(0, np.nan)

# Range ratio — how wide is the spread relative to mean
canopy_2017['I_string_range_ratio'] = (
    (canopy_2017['I_string_max'] - canopy_2017['I_string_min']) / canopy_2017['I_string_mean'].replace(0, np.nan)
)

# 2. Per-String Deviation from Mean
# Z-score deviation identifies WHICH string is the outlier.
# How many standard deviations each string is from the fleet mean.
# This helps identify which string is the outlier, not just that one exists.
# Essential for generating actionable maintenance alerts.
for i in range(1, N_STRINGS + 1):
    col = f'ShuntCurrent_A_Avg_{i}'
    canopy_2017[f'I_string_{i}_dev'] = (
        (canopy_2017[col] - canopy_2017['I_string_mean']) / canopy_2017['I_string_std'].replace(0, np.nan))

# 3. String Power Aggregates
canopy_2017['P_string_mean']  = canopy_2017[string_power_cols].mean(axis=1)
canopy_2017['P_string_std']   = canopy_2017[string_power_cols].std(axis=1)
canopy_2017['P_string_total'] = canopy_2017[string_power_cols].sum(axis=1)

# Power CV
# If CV_power >> CV_current, voltage differences are also present
canopy_2017['P_string_cv'] = canopy_2017['P_string_std'] / canopy_2017['P_string_mean'].replace(0, np.nan)


# 4. Operating Voltage (Ri = Pi / Ii)
# Physics: for a DC string, P = V × I, therefore V = P/I = Ri
# Units: (kW × 1000) / A = W/A = V

for i in range(1, N_STRINGS + 1):
    I_col = f'ShuntCurrent_A_Avg_{i}'
    P_col = f'ShuntPDC_kW_Avg_{i}'
    canopy_2017[f'R_string_{i}'] = (
        (canopy_2017[P_col] * 1000)
        / canopy_2017[I_col].replace(0, np.nan)
    )

# Fleet-level Ri statistics
R_cols = [f'R_string_{i}' for i in range(1, N_STRINGS + 1)]
canopy_2017['R_string_mean'] = canopy_2017[R_cols].mean(axis=1)
canopy_2017['R_string_std']  = canopy_2017[R_cols].std(axis=1)
canopy_2017['R_string_cv']   = (
    canopy_2017['R_string_std']
    / canopy_2017['R_string_mean'].replace(0, np.nan)
)

# Per-string Ri deviation from fleet mean
# Low Ri_dev (strongly negative z-score) = voltage collapsed on that string
for i in range(1, N_STRINGS + 1):
    canopy_2017[f'R_string_{i}_dev'] = (
        (canopy_2017[f'R_string_{i}'] - canopy_2017['R_string_mean'])
        / canopy_2017['R_string_std'].replace(0, np.nan)
    )


# 5. POA-Normalised Features
# Removes irradiance-driven variation, isolating string-specific behaviour.
# Persistent drop in Pnorm or Inorm for one string = fault or degradation.

for i in range(1, N_STRINGS + 1):
    I_col = f'ShuntCurrent_A_Avg_{i}'
    P_col = f'ShuntPDC_kW_Avg_{i}'

    canopy_2017[f'Pnorm_string_{i}'] = canopy_2017[P_col] / canopy_2017['POA_fused'].replace(0, np.nan)
    canopy_2017[f'Inorm_string_{i}'] = canopy_2017[I_col] / canopy_2017['POA_fused'].replace(0, np.nan)

pnorm_cols = [f'Pnorm_string_{i}' for i in range(1, N_STRINGS + 1)]
canopy_2017['Pnorm_string_std'] = canopy_2017[pnorm_cols].std(axis=1)
canopy_2017['Pnorm_string_cv']  = canopy_2017['Pnorm_string_std'] / canopy_2017[pnorm_cols].mean(axis=1).replace(0, np.nan)


# 6. Rolling CV for Sustained Detection
canopy_2017['I_string_cv_roll5'] = canopy_2017['I_string_cv'].rolling(window=5, min_periods=3).mean()

# 6. Per-String Baseline Normalisation
# The Canopy Array has structural current asymmetry, combiner boxes 1 and 7 serve fewer source circuits than 2–4.
# Raw CV will chronically exceed 0.15 even during healthy operation.
#
# Correct approach: compare each string to its own irradiance-adjusted baseline, not to the fleet mean.
# A string's expected current scales linearly with POA (ISC ∝ G).
# The median of Inorm_i over evaluable hours gives each string's individual characteristic response under healthy conditions.
evaluable_baseline = ((canopy_2017['Op_Mask'] == 1) & (canopy_2017['POA_fused'] >= POA_MIN_STRING) &
                               (canopy_2017['InvOpStatus_Avg'] > 0) & (canopy_2017['InvOpStatus_Avg'] <= 5))

INORM_BASELINE = {}
for i in range(1, N_STRINGS + 1):
    INORM_BASELINE[i] = canopy_2017.loc[evaluable_baseline, f'Inorm_string_{i}'].median()

print("Per-string Inorm baselines (healthy fingerprint):")
for i, v in INORM_BASELINE.items():
    print(f"  String {i}: {v:.6f} A·m²/W")

# Expected current for each string at current POA, using its own baseline
for i in range(1, N_STRINGS + 1):
    canopy_2017[f'I_string_{i}_expected'] = INORM_BASELINE[i] * canopy_2017['POA_fused']
    # Relative deviation: how far is actual current from this string's own expectation?
    # Negative = underproducing relative to its own history → fault candidate
    # Positive = overproducing (unusual, could indicate measurement drift)
    canopy_2017[f'I_string_{i}_rel_dev'] = (
        (canopy_2017[f'ShuntCurrent_A_Avg_{i}'] - canopy_2017[f'I_string_{i}_expected'])
        / canopy_2017[f'I_string_{i}_expected'].replace(0, np.nan)
    )

# Fleet-level imbalance using baseline-corrected deviations
# Max absolute relative deviation across all strings — one large outlier signals imbalance
rel_dev_cols = [f'I_string_{i}_rel_dev' for i in range(1, N_STRINGS + 1)]
canopy_2017['I_rel_dev_max'] = canopy_2017[rel_dev_cols].abs().max(axis=1)
canopy_2017['I_rel_dev_std'] = canopy_2017[rel_dev_cols].std(axis=1)

# Rolling versions for sustained detection
canopy_2017['I_rel_dev_max_roll5'] = canopy_2017['I_rel_dev_max'].rolling(window=5, min_periods=3).mean()
````

````python
# 7. String Fault Labels

# Threshold: a string deviating > 20% below its own baseline
# for ≥ 5 minutes is flagged as imbalance.
# 20% chosen to be robust to normal irradiance transients (~5%)
# and sensor noise (~2%), while catching real faults (soiling
# typically causes 10–40% loss, string faults cause 14–100%).
STRING_REL_DEV_THRESHOLD = 0.20

evaluable = evaluable_baseline

# Open circuit: one string carrying < 10% of fleet mean while system is producing
# Use I_string_min to find the lowest string — if it's nearly zero, one string is out
open_circuit = (
    evaluable
    & (canopy_2017['I_string_min'] < OPEN_CIRCUIT_RATIO * canopy_2017['I_string_mean'])
)

# String Imbalance (baseline-normalised)
# Fires when ANY string deviates > 20% below its own expected current, sustained over a 5-minute rolling window.
# This is robust to the structural asymmetry between combiner boxes because each string is compared to its own characteristic, not the fleet mean.
string_imbalance = (
    evaluable
    & ~open_circuit
    & (canopy_2017['I_rel_dev_max_roll5'] > STRING_REL_DEV_THRESHOLD)
)

conditions_B = [
    ~evaluable,
    canopy_2017['I_string_mean'].isna(),
    open_circuit,
    string_imbalance,
]
choices_B = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'OPEN_CIRCUIT_STRING',
    'STRING_IMBALANCE',
]
canopy_2017['label_string'] = np.select(conditions_B, choices_B, default='NORMAL')

# Check
print("=== Labeling B: String-Level Features ===\n")
ev = canopy_2017[evaluable]

print("String current stats (after bounds cleaning):")
print(canopy_2017[string_current_cols].describe().round(3).to_string())

print(f"\nPer-string mean current and Ri during daytime (POA ≥ {POA_MIN_STRING}):")
for i in range(1, N_STRINGS + 1):
    col = f'ShuntCurrent_A_Avg_{i}'
    mean_i  = canopy_2017.loc[evaluable, col].mean()
    mean_r  = canopy_2017.loc[evaluable, f'R_string_{i}'].mean()
    baseline = INORM_BASELINE[i]
    print(f"  String {i}: I = {mean_i:.2f} A   Ri = {mean_r:.1f} V   Inorm_baseline = {baseline:.5f}")

print(f"\nBaseline-corrected relative deviation (I_rel_dev) distribution (daytime):")
for i in range(1, N_STRINGS + 1):
    col = f'I_string_{i}_rel_dev'
    med = canopy_2017.loc[evaluable, col].median()
    p5  = canopy_2017.loc[evaluable, col].quantile(0.05)
    p95 = canopy_2017.loc[evaluable, col].quantile(0.95)
    print(f"  String {i}: median = {med:+.3f}   p5 = {p5:+.3f}   p95 = {p95:+.3f}")

print(f"\nRi (voltage proxy) distribution (daytime):")
print(ev[R_cols].describe().round(2).to_string())

print(f"\nlabel_string distribution:")
print(canopy_2017['label_string'].value_counts())
print(f"\nAs % of total rows:")
print((canopy_2017['label_string'].value_counts() / len(canopy_2017) * 100).round(2))
````

# Inverter Level Faults

````python
canopy_2017 = pd.read_csv('/content/drive/MyDrive/NIST_processed/canopy_2017_Label_2.csv', index_col='TIMESTAMP')
````

````python
POA_MIN_INVERTER        = 100    # W/m² — threshold for clear daylight
INV_SHUTDOWN_POWER      = 0.05   # kW — below this = inverter not producing
INV_EFFICIENCY_MIN      = 0.85   # DC→AC efficiency floor for healthy inverter
INV_EFFICIENCY_MAX      = 1.00   # ceiling — above this is physically impossible
INV_RATED_DC_KW         = 243.0  # system DC nameplate
INV_RATED_AC_KW         = 260.0  # inverter AC rated output

# 1. DC → AC Conversion Efficiency
# η = PAC / PDC
# Healthy inverter: 0.93–0.98 during normal operation
# Drops toward zero during startup/shutdown (normal)
# Sustained drop below 0.85 during full operation = inverter fault
# Above 1.0 is impossible
canopy_2017['inv_efficiency'] = (
    canopy_2017['InvPAC_kW_Avg'] / canopy_2017['InvPDC_kW_Avg'].replace(0, np.nan)
)
# Clip to physically meaningful range
canopy_2017['inv_efficiency_clipped'] = canopy_2017['inv_efficiency'].clip(lower=0, upper=INV_EFFICIENCY_MAX)


# 2. DC Power from Strings vs Inverter DC Input
# P_string_total = sum of all 7 shunt-measured string powers
# InvPDC_kW_Avg = inverter's own DC input measurement
# Discrepancy between the two = wiring loss, measurement error, or DC fault between the combiner box and inverter input
canopy_2017['DC_power_discrepancy'] = canopy_2017['P_string_total'] - canopy_2017['InvPDC_kW_Avg']

canopy_2017['DC_power_discrepancy_pct'] = canopy_2017['DC_power_discrepancy'] / canopy_2017['InvPDC_kW_Avg'].replace(0, np.nan) * 100


# 3. Inverter Loading Ratio
# How hard is the inverter working relative to its AC rating?
# Clipping occurs when PDC >> AC rating (common on sunny winter days)
# Persistent overloading or underloading relative to available DC = fault
canopy_2017['inv_loading_ratio'] = canopy_2017['InvPAC_kW_Avg'] / INV_RATED_AC_KW


# 4. Inverter Fault Flag Composite
# The inverter reports its own fault codes.
fault_flag_cols = [
    'InvMainFault_Max', 'InvDriveFault_Max', 'InvVoltageFault_Max',
    'InvGridFault_Max', 'InvTempFault_Max', 'InvSystemFault_Max'
]
# Any non-zero fault code = inverter self-reported fault
canopy_2017['inv_any_fault_reported'] = (canopy_2017[fault_flag_cols] > 0).any(axis=1).astype(int)
canopy_2017['inv_fault_count']        = (canopy_2017[fault_flag_cols] > 0).sum(axis=1)


# 5. Rolling Efficiency
canopy_2017['inv_efficiency_roll5'] = canopy_2017['inv_efficiency_clipped'].rolling(window=5, min_periods=3).mean()
````

````python
canopy_2017 = canopy_2017.copy()
````

````python
evaluable_inv = (canopy_2017['Op_Mask'] == 1) & (canopy_2017['POA_fused'] >= POA_MIN_INVERTER)

# Inverter Shutdown
# Two distinct shutdown conditions, both require Op_Mask = 1:
#
# Condition A — Power-based: PAC < 0.05 kW despite sufficient irradiance.
#   The inverter is not exporting power. Root causes: protection trip,
#   thermal shutdown, manual disconnect, communication fault.
#
# Condition B — Status-based: InvOpStatus_Avg <= 0 during operational hours.
#   The inverter self-reports it is not in an active operating state.
#   From the PVP260kW inverter (Figure B-2, NIST TN 1896), status codes
#   ≤ 0 indicate fault, standby, or off states. Status 1–5 covers normal
#   operating modes (MPPT tracking, startup, derating).
#   This catches shutdowns the inverter reports but power measurement
#   alone might miss during brief transitions.
#
# Note: Condition B rows are also NOT_EVALUABLE for efficiency analysis
# when the inverter is not in normal operation, η = PAC/PDC is meaningless.

inverter_shutdown_power = evaluable_inv & (canopy_2017['InvPAC_kW_Avg'] < INV_SHUTDOWN_POWER)

inverter_shutdown_status = (
    evaluable_inv
    & (canopy_2017['InvOpStatus_Avg'].isna() | (canopy_2017['InvOpStatus_Avg'] <= 0))
)

inverter_shutdown = inverter_shutdown_power | inverter_shutdown_status

# Inverter Efficiency Fault
# Requires:
#   - InvOpStatus in normal operating range (1–5): inverter is actually running
#   - Not already caught as shutdown
#   - Sustained η < 0.85 over 5-minute rolling window
#   - PDC > 1.0 kW: efficiency undefined at near-zero load
#   - Loading > 20%: normal efficiency curves dip at light load
#
# The InvOpStatus guard is critical — without it, startup and shutdown
# transitions produce brief low-efficiency readings that fire false positives.

inv_normal_operation = (canopy_2017['InvOpStatus_Avg'] > 0) & (canopy_2017['InvOpStatus_Avg'] <= 5)

inv_efficiency_fault = (
    evaluable_inv
    & inv_normal_operation
    & ~inverter_shutdown
    & (canopy_2017['inv_efficiency_roll5'] < INV_EFFICIENCY_MIN)
    & (canopy_2017['InvPDC_kW_Avg'] > 1.0)
    & (canopy_2017['inv_loading_ratio'] > 0.20)
)

# Label Assignment
# Priority: NOT_EVALUABLE > MISSING_DATA > INVERTER_SHUTDOWN >
#           INVERTER_EFFICIENCY_FAULT > NORMAL

conditions_C = [
    canopy_2017['Op_Mask'] == 0,
    canopy_2017['InvPAC_kW_Avg'].isna() | canopy_2017['InvPDC_kW_Avg'].isna(),
    inverter_shutdown,
    inv_efficiency_fault,
]
choices_C = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'INVERTER_SHUTDOWN',
    'INVERTER_EFFICIENCY_FAULT',
]
canopy_2017['label_inverter'] = np.select(conditions_C, choices_C, default='NORMAL')

# Store shutdown sub-type for analysis — useful for cross-checking
# which mechanism is catching more events
canopy_2017['inv_shutdown_power_flag']  = inverter_shutdown_power.astype(int)
canopy_2017['inv_shutdown_status_flag'] = inverter_shutdown_status.astype(int)

# Check
print("=== Labeling C: Inverter-Level Features ===\n")

eff_mask = evaluable_inv & inv_normal_operation & (canopy_2017['InvPDC_kW_Avg'] > 1.0)

print("DC→AC efficiency (daytime, normal operation, PDC > 1 kW):")
print(canopy_2017.loc[eff_mask, 'inv_efficiency'].describe().round(4))

print(f"\nEfficiency ceiling violations (η > 1.0):")
above_one = (canopy_2017.loc[eff_mask, 'inv_efficiency'] > 1.0).sum()
print(f"  Count : {above_one} rows ({above_one / eff_mask.sum() * 100:.2f}% of evaluable)")

print(f"\nShutdown breakdown:")
print(f"  Power-based (PAC < {INV_SHUTDOWN_POWER} kW)   : {inverter_shutdown_power.sum()}")
print(f"  Status-based (OpStatus ≤ 0)  : {inverter_shutdown_status.sum()}")
print(f"  Both simultaneously          : {(inverter_shutdown_power & inverter_shutdown_status).sum()}")
print(f"  Total unique shutdown rows   : {inverter_shutdown.sum()}")

print(f"\nDC power discrepancy % (daytime, normal operation):")
print(canopy_2017.loc[eff_mask, 'DC_power_discrepancy_pct'].describe().round(3))

print(f"\nInverter loading ratio (daytime, normal operation):")
print(canopy_2017.loc[eff_mask, 'inv_loading_ratio'].describe().round(3))

print(f"\nInverter self-reported faults:")
print(f"  Rows with any fault flag > 0 : {canopy_2017['inv_any_fault_reported'].sum()}")
print(f"  Max simultaneous fault codes : {canopy_2017['inv_fault_count'].max()}")
per_flag = (canopy_2017[fault_flag_cols] > 0).sum()
print(f"  Per-flag breakdown:\n{per_flag.to_string()}")

print(f"\nlabel_inverter distribution:")
print(canopy_2017['label_inverter'].value_counts())
print(f"\nAs % of total rows:")
print((canopy_2017['label_inverter'].value_counts() / len(canopy_2017) * 100).round(2))

print(f"\nCross-check — physics labels vs inverter self-reported faults:")
shutdown_rows   = canopy_2017['label_inverter'] == 'INVERTER_SHUTDOWN'
eff_fault_rows  = canopy_2017['label_inverter'] == 'INVERTER_EFFICIENCY_FAULT'
reported        = canopy_2017['inv_any_fault_reported'].astype(bool)

print(f"  Shutdown detected by physics           : {shutdown_rows.sum()}")
print(f"    → power-based only                   : {(canopy_2017['inv_shutdown_power_flag']==1 & ~canopy_2017['inv_shutdown_status_flag'].astype(bool)).sum()}")
print(f"    → status-based only                  : {(canopy_2017['inv_shutdown_status_flag']==1 & ~canopy_2017['inv_shutdown_power_flag'].astype(bool)).sum()}")
print(f"    → also self-reported by inverter     : {(shutdown_rows & reported).sum()}")
print(f"    → NOT self-reported (silent fault?)  : {(shutdown_rows & ~reported).sum()}")
print(f"  Efficiency fault detected by physics   : {eff_fault_rows.sum()}")
print(f"    → also self-reported by inverter     : {(eff_fault_rows & reported).sum()}")
print(f"    → NOT self-reported                  : {(eff_fault_rows & ~reported).sum()}")
print(f"  Self-reported only (physics missed)    : {(~shutdown_rows & ~eff_fault_rows & reported & evaluable_inv).sum()}")
````
