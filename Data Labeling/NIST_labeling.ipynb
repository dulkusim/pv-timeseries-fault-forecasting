In this step, a correlation matrix is computed for a selected group of inverter and power meter electrical measurements to evaluate potential redundancy between features.

The goal is to identify whether inverter-side measurements and power meter measurements provide overlapping information and could therefore be removed to reduce dimensionality.

The selected features include:

- Inverter AC/DC voltages and currents
- Power meter measurements
- Grid frequency and power output signals

Correlation was computed using:

```python
similars.corr()
```

---

### Key Observation

The analysis shows **strong correlations between inverter and power meter variables**, particularly for currents and power-related measurements.
  - `InvIa/Ib/Ic` and `PwrMtrIa/Ib/Ic`
  - `InvPAC_kW_Avg` and `PwrMtrP_kW_Avg`

However, the correlation values between the most similar pairs are typically around:

- ~0.86–0.87 for key power/current relationships

While this indicates a strong linear relationship, it is **below a redundancy threshold of ≈ 0.9** for feature elimination.

---

### Decision

Although the features are highly correlated, they were **not removed** because:

- The correlation is strong but not extreme (< 0.9)
- Power meter and inverter measurements may still capture different physical effects (e.g., grid-side losses, measurement noise, conversion efficiency)
- Removing them at this stage could lead to loss of potentially useful signal for downstream modeling

---

### Conclusion

Despite noticeable redundancy between inverter and power meter features, the correlation is not sufficiently high to justify removal. Therefore, all selected features are retained for further analysis and future model-based feature selection.

## Operating Mask Creation

In this step, an operating mask is created to distinguish periods where the photovoltaic system is actively producing power from periods with little or no solar irradiance.

The median value of the three pyranometer measurements is used as a robust estimate of solar irradiance:

```python
median_irradiance = canopy_2015[pyra_cols].median(axis=1)
```

Using the median helps reduce the impact of noisy or faulty irradiance sensors.

An operating mask (`Op_Mask`) is then generated using a threshold of:

```text
100 W/m²
```

- `1` → System considered operational (sufficient irradiance)
- `0` → Low irradiance or nighttime conditions

This filtering step is important because photovoltaic systems naturally produce near-zero power during nighttime or very low irradiance periods, which should not be treated as faults during later analysis and modeling.

````python
pyra_cols = ['Pyra1_Wm2_Avg', 'Pyra2_Wm2_Avg', 'Pyra3_Wm2_Avg']

median_irradiance = canopy_2015[pyra_cols].median(axis=1)

canopy_2015['Op_Mask'] = np.where(median_irradiance > 100, 1, 0)
canopy_2015
````

## Expected vs Actual Power Analysis (Performance Ratio)

In this step, physics-based features are engineered to compare the photovoltaic system’s expected power production against its actual inverter output.

The expected power is estimated using:

- Plane-of-array irradiance from fused reference cell measurements
- Module temperature
- PV system rated capacity
- Temperature correction coefficient

The implemented model estimates the theoretical power that the system should produce under current environmental conditions.

Several diagnostic features are then created, including:

- `P_expected_kW` → Theoretical expected PV power
- `PR` (Performance Ratio) → Actual power / Expected power
$$
P_{\text{expected}} = P_{\text{rated}} \times \left( \frac{\text{POA}_{\text{fused}}}{1000} \right) \times \left[ 1 + \gamma \times (T_{\text{mod}} - T_{\text{STC}}) \right]
$$
- `power_deficit_kW` → Difference between expected and actual power
- Rolling PR statistics → Short-term temporal behavior of system efficiency
- `PR_30d_trend_pct` → Long-term degradation trend indicator

Additional irradiance consistency features are also generated:

- Sensor spread between reference cells
- Coefficient of variation between irradiance sensors

The purpose of this feature engineering step is to create physically meaningful indicators for:

- PV underperformance detection
- Fault analysis
- System degradation monitoring
- Time-series anomaly detection
- Machine learning model training

````python
canopy_2015 = pd.read_csv('/content/drive/MyDrive/NIST_processed/canopy_2015_clean.csv')
````

````python
# Convert the TIMESTAMP column to pandas datetime object
canopy_2015['TIMESTAMP'] = pd.to_datetime(canopy_2015['TIMESTAMP'])

# Set TIMESTAMP as index
canopy_2015.set_index('TIMESTAMP', inplace=True)
````

````python
P_RATED_DC_KW  = 243.0    # DC nameplate capacity (kW) — NIST Canopy
P_AC_MAX_KW    = 225.0    # Inverter AC rated output (~0.93 × DC nameplate)
GAMMA          = -0.004   # Power temperature coefficient for monocrystalline Si (/°C)
T_STC          = 25.0     # Standard Test Condition reference temperature (°C)

# IEC 61724-1 minimum irradiance for valid PR calculations
# Below this threshold, measurement noise dominates relative errors
# This is SEPARATE from Op_Mask (100 W/m²) which defines operational hours
POA_MIN_PR     = 200.0    # W/m² — IEC 61724-1


# 1. POA Irradiance
# Using only the two reference cells (spectral response matches modules)
# rather than mixing with flat-spectral pyranometers.
# Median is robust to one sensor being soiled or faulty.
irrad_cols = ['RefCell1_Wm2_Avg', 'RefCell2_Wm2_Avg']
canopy_2015['POA_fused'] = canopy_2015[irrad_cols].median(axis=1)

# Sensor agreement metrics - high spread means sensors disagree (soiling, partial shade, fault)
canopy_2015['irrad_sensor_spread'] = canopy_2015[irrad_cols].max(axis=1) - canopy_2015[irrad_cols].min(axis=1)
canopy_2015['irrad_sensor_cv']     = canopy_2015[irrad_cols].std(axis=1) / canopy_2015[irrad_cols].mean(axis=1).replace(0, np.nan)

# IEC 61724-1 validity flag — PR calculations only trustworthy above 200 W/m²
canopy_2015['PR_evaluable'] = ((canopy_2015['Op_Mask'] == 1) & (canopy_2015['POA_fused'] >= POA_MIN_PR)).astype(int)


# 2. Temperature-Corrected Expected Power
# Standard IEC 61724-1:2021:
# P_expected = P_rated × (poa_fused / 1000) × [1 + γ × (T_mod - T_STC)]
# This is the power the system should be producing given current irradiance and module temperature.
# Shortfalls indicate a fault signal.

canopy_2015['module_temp'] = canopy_2015['SEWSModuleTemp_C_Avg'] #for readability

canopy_2015['P_expected_DC_kW'] = (P_RATED_DC_KW * (canopy_2015['POA_fused'] / 1000.0) * (1 + GAMMA * (canopy_2015['module_temp'] - T_STC)))


# 3. AC-Bounded Expected Power
# The numerator of PR is AC output, but the raw P_expected_DC can exceed the inverter's AC ceiling on high-irradiance days.

canopy_2015['P_expected_AC_kW'] = canopy_2015['P_expected_DC_kW'].clip(upper=P_AC_MAX_KW)


# 4. Performance Ratio
# PR = P_actual_AC / P_expected
#
# Threshold justification (all based on Op_Mask=1 AND POA ≥ 200 W/m²):
#
#   > 1.05    : Suspiciously high — measurement artifact or transient cloud edge.
#
#   0.85–1.05 : NORMAL — within expected real-world loss range.
#
#   0.70–0.85 : UNDERPERFORMANCE — measurable degradation beyond normal losses.
#
#   0.50–0.70 : SEVERE UNDERPERFORMANCE — likely 1–2 string outages, significant soiling event, or partial inverter fault.
#               1 string out of 7 → ~14% loss → PR drops from 0.82 to ~0.71.
#               2 strings → ~28% loss → PR ≈ 0.59. Threshold placement is
#
#   0.00–0.50 : CRITICAL UNDERPERFORMANCE — near-total production loss during operational hours.
#               Consistent with inverter shutdown or 3+ strings out.
#
#   < 0       : NEGATIVE AC POWER — inverter drawing from grid (standby/housekeeping).
#               Normal at night; abnormal during Op_Mask=1 hours.

canopy_2015['PR'] = canopy_2015['InvPAC_kW_Avg'] / canopy_2015['P_expected_AC_kW'].replace(0, np.nan)


# 5. Energy-Weighted Rolling PR
# Plain rolling mean of PR gives equal weight to low- and high-irradiance minutes.
# At low irradiance, small absolute errors produce large relative PR noise.
# The energy-weighted formulation matches IEC 61724-1 period-level PR definition:
#   PR_weighted = Σ(P_actual) / Σ(P_expected)  over the window
#
# This naturally down-weights low-production minutes because they contribute less to both numerator and denominator in absolute terms.
# Single consistent window (15 min) used for all thresholds — eliminates the inconsistency of using different windows for different severity levels.

canopy_2015['PR_roll_15min_weighted'] = canopy_2015['InvPAC_kW_Avg'].rolling(window=15, min_periods=8).sum() / canopy_2015['P_expected_AC_kW'].rolling(window=15, min_periods=8).sum()

# Additional windows for feature richness (used as ML features, not for labels)
canopy_2015['PR_roll_5min_weighted'] = canopy_2015['InvPAC_kW_Avg'].rolling(window=5, min_periods=3).sum() / canopy_2015['P_expected_AC_kW'].rolling(window=5, min_periods=3).sum()

canopy_2015['PR_roll_60min_weighted'] = canopy_2015['InvPAC_kW_Avg'].rolling(window=60, min_periods=30).sum() / canopy_2015['P_expected_AC_kW'].rolling(window=60, min_periods=30).sum()

# Rolling volatility of PR — spiky PR = intermittent fault; sustained low = chronic fault
canopy_2015['PR_roll_std_15min'] = canopy_2015['PR'].rolling(window=15, min_periods=8).std()

# Power deficit — absolute kW gap, useful for severity ranking in labeling merge
canopy_2015['power_deficit_kW'] = canopy_2015['P_expected_AC_kW'] - canopy_2015['InvPAC_kW_Avg']

# 6. Degradation Signal (30-day rolling trend)
# We want to detect PR trending downward over time.
# Method: 30-day rolling mean of PR, then compute % change from the value 30 days before.

# Downsample to hourly for degradation trend (degradation is a slow process, 1-min resolution adds nothing)
pr_hourly = canopy_2015['PR'].resample('1h').mean()

# 30-day rolling mean on hourly series (720 hours = 30 days)
pr_30d_roll = pr_hourly.rolling(window=720, min_periods=360).mean()

# % change vs 30 days ago
pr_30d_pct_change = pr_30d_roll.ffill().pct_change(periods=720) * 100

# Reindex back to 1-min — forward fill within each hour (same degradation rate)
canopy_2015['PR_30d_trend_pct'] = pr_30d_pct_change.reindex(canopy_2015.index, method='ffill')
````

````python
round_cols = [
    'irrad_sensor_spread', 'irrad_sensor_cv', 'P_expected_DC_kW', 'P_expected_AC_kW',
    'PR', 'PR_roll_5min_weighted', 'PR_roll_15min_weighted', 'PR_roll_60min_weighted',
    'PR_roll_std_15min', 'power_deficit_kW', 'PR_30d_trend_pct'
]

canopy_2015[round_cols] = canopy_2015[round_cols].round(4)
````

## Physics-Based Fault Labeling

In this step, rule-based fault labels are generated using the previously engineered expected-vs-actual power features.

The labeling process is based primarily on the photovoltaic system’s Performance Ratio (PR), which compares actual inverter output against the theoretically expected power under current irradiance and temperature conditions.

The following labels are assigned:

| Label | Description |
|---|---|
| `NOT_EVALUABLE` | Nighttime or very low irradiance conditions (`Op_Mask = 0`) |
| `MISSING_DATA` | Missing critical measurements required for evaluation |
| `SEVERE_UNDERPERFORMANCE` | Severe sustained underperformance (`PR_roll_5min < 0.50`) |
| `UNDERPERFORMANCE` | Moderate sustained underperformance (`PR_roll_15min < 0.70`) |
| `NORMAL` | System operating within expected performance range |

A priority hierarchy is also applied:

```text
SEVERE_UNDERPERFORMANCE > UNDERPERFORMANCE > NORMAL
```

Rolling window statistics are used instead of instantaneous PR values in order to reduce noise and avoid labeling short transient fluctuations as faults.

This approach creates weak physics-based labels that can later be used for:

- Fault detection experiments
- Supervised machine learning
- Anomaly detection validation
- PV system performance monitoring

````python
#Fault Labeling

# Priority order: NOT_EVALUABLE > MISSING_DATA > NEGATIVE_AC_POWER > CRITICAL_UNDERPERFORMANCE > SEVERE_UNDERPERFORMANCE >
#                 UNDERPERFORMANCE > SUSPICIOUSLY_HIGH > NORMAL

pr = canopy_2015['PR_roll_15min_weighted']
evaluable = canopy_2015['PR_evaluable'] == 1

conditions = [
    # 1. Outside operating window or below irradiance floor
    canopy_2015['PR_evaluable'] == 0,

    # 2. Cannot compute PR — critical input is missing
    canopy_2015['P_expected_AC_kW'].isna() | canopy_2015['InvPAC_kW_Avg'].isna(),

    # 3. Negative AC power during operational hours
    # Inverter drawing from grid — abnormal during daylight
    evaluable & (canopy_2015['InvPAC_kW_Avg'] < 0),

    # 4. Critical underperformance: PR 0.00–0.50
    # Near-total production loss — inverter shutdown or 3+ strings out
    evaluable & (pr >= 0.0) & (pr < 0.50),

    # 5. Severe underperformance: PR 0.50–0.70
    # Consistent with 1–2 string outages or significant partial shading
    evaluable & (pr >= 0.50) & (pr < 0.70),

    # 6. Underperformance: PR 0.70–0.85
    # Below realistic expectation for healthy c-Si system
    evaluable & (pr >= 0.70) & (pr < 0.85),

    # 7. Suspiciously high: PR > 1.05
    # Physically possible but rare
    evaluable & (pr > 1.05),
]

choices = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'NEGATIVE_AC_POWER',
    'CRITICAL_UNDERPERFORMANCE',
    'SEVERE_UNDERPERFORMANCE',
    'UNDERPERFORMANCE',
    'SUSPICIOUSLY_HIGH',
]

canopy_2015['label_power'] = np.select(conditions, choices, default='NORMAL')


# Check results
print("=== Labeling: Power Performance ===\n")

print("PR distribution (PR_evaluable = 1 rows only):")
print(canopy_2015.loc[evaluable, 'PR'].describe().round(4))

print(f"\nWeighted rolling PR distribution (PR_evaluable = 1):")
print(canopy_2015.loc[evaluable, 'PR_roll_15min_weighted'].describe().round(4))

print(f"\nlabel_power distribution:")
print(canopy_2015['label_power'].value_counts())
print(f"\nAs percentage of total rows:")
print((canopy_2015['label_power'].value_counts() / len(canopy_2015) * 100).round(2))

print(f"\nPR_evaluable breakdown:")
print(f"  Op_Mask=1 rows total          : {(canopy_2015['Op_Mask']==1).sum()}")
print(f"  PR_evaluable=1 (POA ≥ 200)   : {(canopy_2015['PR_evaluable']==1).sum()}")
print(f"  Rows lost to IEC floor        : {((canopy_2015['Op_Mask']==1) & (canopy_2015['PR_evaluable']==0)).sum()}")

print(f"\nNEGATIVE_AC_POWER during PR_evaluable hours:")
neg = (canopy_2015['label_power'] == 'NEGATIVE_AC_POWER')
print(f"  Count: {neg.sum()} rows ({neg.sum()/60:.1f} hours)")
````

##String Level Features

````python
string_current_cols = [f'ShuntCurrent_A_Avg_{i}' for i in range(1, 8)]
string_power_cols   = [f'ShuntPDC_kW_Avg_{i}'    for i in range(1, 8)]

# Apply physical bounds to string currents
# PV string current cannot be negative or exceed module Isc * parallel strings
CURRENT_MIN = 0.0    # A — negative current is physically impossible for DC strings
CURRENT_MAX = 150.0  # A — ceiling above observed max of ~139A

for col in string_current_cols:
    canopy_2015[col] = canopy_2015[col].where(
        canopy_2015[col].between(CURRENT_MIN, CURRENT_MAX), other=np.nan
    )

# Same for string power
POWER_MIN = 0.0
POWER_MAX = 50.0    # kW per string — 243kW / 7 strings ≈ 34.7 kW max
for col in string_power_cols:
    canopy_2015[col] = canopy_2015[col].where(
        canopy_2015[col].between(POWER_MIN, POWER_MAX), other=np.nan
    )
````

````python
POA_MIN_STRING = 300      # W/m² — minimum irradiance to evaluate string faults
N_STRINGS      = 7
STRING_CV_THRESHOLD       = 0.15   # coefficient of variation threshold
OPEN_CIRCUIT_RATIO        = 0.10   # string is "dead" if < 10% of mean
SHORT_CIRCUIT_RI_RATIO    = 0.40   # operating voltage proxy < X% of fleet mean


# 1. String Current Aggregates
canopy_2015['I_string_mean']  = canopy_2015[string_current_cols].mean(axis=1)
canopy_2015['I_string_std']   = canopy_2015[string_current_cols].std(axis=1)
canopy_2015['I_string_min']   = canopy_2015[string_current_cols].min(axis=1)
canopy_2015['I_string_max']   = canopy_2015[string_current_cols].max(axis=1)

# Coefficient of variation — spread relative to mean, scale-independent
# High CV = strings behaving differently from each other
canopy_2015['I_string_cv'] = canopy_2015['I_string_std'] / canopy_2015['I_string_mean'].replace(0, np.nan)

# Range ratio — how wide is the spread relative to mean
canopy_2015['I_string_range_ratio'] = (
    (canopy_2015['I_string_max'] - canopy_2015['I_string_min']) / canopy_2015['I_string_mean'].replace(0, np.nan)
)

# 2. Per-String Deviation from Mean
# Z-score deviation identifies WHICH string is the outlier.
# How many standard deviations each string is from the fleet mean.
# This helps identify which string is the outlier, not just that one exists.
# Essential for generating actionable maintenance alerts.
for i in range(1, N_STRINGS + 1):
    col = f'ShuntCurrent_A_Avg_{i}'
    canopy_2015[f'I_string_{i}_dev'] = (
        (canopy_2015[col] - canopy_2015['I_string_mean']) / canopy_2015['I_string_std'].replace(0, np.nan))

# 3. String Power Aggregates
canopy_2015['P_string_mean']  = canopy_2015[string_power_cols].mean(axis=1)
canopy_2015['P_string_std']   = canopy_2015[string_power_cols].std(axis=1)
canopy_2015['P_string_total'] = canopy_2015[string_power_cols].sum(axis=1)

# Power CV
# If CV_power >> CV_current, voltage differences are also present
canopy_2015['P_string_cv'] = canopy_2015['P_string_std'] / canopy_2015['P_string_mean'].replace(0, np.nan)


# 4. Operating Voltage (Ri = Pi / Ii)
# Physics: for a DC string, P = V × I, therefore V = P/I = Ri
# Units: (kW × 1000) / A = W/A = V

for i in range(1, N_STRINGS + 1):
    I_col = f'ShuntCurrent_A_Avg_{i}'
    P_col = f'ShuntPDC_kW_Avg_{i}'
    canopy_2015[f'R_string_{i}'] = (
        (canopy_2015[P_col] * 1000)
        / canopy_2015[I_col].replace(0, np.nan)
    )

# Fleet-level Ri statistics
R_cols = [f'R_string_{i}' for i in range(1, N_STRINGS + 1)]
canopy_2015['R_string_mean'] = canopy_2015[R_cols].mean(axis=1)
canopy_2015['R_string_std']  = canopy_2015[R_cols].std(axis=1)
canopy_2015['R_string_cv']   = (
    canopy_2015['R_string_std']
    / canopy_2015['R_string_mean'].replace(0, np.nan)
)

# Per-string Ri deviation from fleet mean
# Low Ri_dev (strongly negative z-score) = voltage collapsed on that string
for i in range(1, N_STRINGS + 1):
    canopy_2015[f'R_string_{i}_dev'] = (
        (canopy_2015[f'R_string_{i}'] - canopy_2015['R_string_mean'])
        / canopy_2015['R_string_std'].replace(0, np.nan)
    )


# 5. POA-Normalised Features
# Removes irradiance-driven variation, isolating string-specific behaviour.
# Persistent drop in Pnorm or Inorm for one string = fault or degradation.

for i in range(1, N_STRINGS + 1):
    I_col = f'ShuntCurrent_A_Avg_{i}'
    P_col = f'ShuntPDC_kW_Avg_{i}'

    canopy_2015[f'Pnorm_string_{i}'] = canopy_2015[P_col] / canopy_2015['POA_fused'].replace(0, np.nan)

    canopy_2015[f'Inorm_string_{i}'] = canopy_2015[I_col] / canopy_2015['POA_fused'].replace(0, np.nan)

pnorm_cols = [f'Pnorm_string_{i}' for i in range(1, N_STRINGS + 1)]
canopy_2015['Pnorm_string_std'] = canopy_2015[pnorm_cols].std(axis=1)
canopy_2015['Pnorm_string_cv']  = canopy_2015['Pnorm_string_std'] / canopy_2015[pnorm_cols].mean(axis=1).replace(0, np.nan)


# 6. Rolling CV for Sustained Detection
canopy_2015['I_string_cv_roll5'] = canopy_2015['I_string_cv'].rolling(window=5, min_periods=3).mean()
````

````python
# 7. String Fault Labels
# Evaluated only where Op_Mask = 1 AND irradiance above threshold.
# At low irradiance, string currents are near zero so CV explodes artificially.

evaluable = (canopy_2015['Op_Mask'] == 1) & (canopy_2015['POA_fused'] >= POA_MIN_STRING)

# Open circuit: one string carrying < 10% of fleet mean while system is producing
# Use I_string_min to find the lowest string — if it's nearly zero, one string is out
open_circuit = (
    evaluable & (canopy_2015['I_string_min'] < OPEN_CIRCUIT_RATIO * canopy_2015['I_string_mean'])
)

# Short Circui
# Condition: one string's Ri < 40% of fleet mean Ri
# Physics: short circuit collapses string voltage → Ri = P/I → 0
# Require the string NOT to be open (I not near zero) so Ri not a div/0.
# Identify which string has the collapsed Ri:
R_min_per_row = canopy_2015[R_cols].min(axis=1)

short_circuit = (
    evaluable
    & ~open_circuit
    & (R_min_per_row < SHORT_CIRCUIT_RI_RATIO * canopy_2015['R_string_mean'])
    & (canopy_2015['I_string_min'] >= OPEN_CIRCUIT_RATIO * canopy_2015['I_string_mean'])
)

# String Imbalance
# Condition: sustained CV > 0.15 with no single-string fault identified
# This is the residual category — catches differential soiling, partial shading, gradual degradation mismatch between strings
string_imbalance = (
    evaluable
    & ~open_circuit
    & ~short_circuit
    & (canopy_2015['I_string_cv_roll5'] > STRING_CV_THRESHOLD)
)

conditions_B = [
    canopy_2015['Op_Mask'] == 0,
    canopy_2015['I_string_mean'].isna(),
    open_circuit,
    short_circuit,
    string_imbalance,
]
choices_B = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'OPEN_CIRCUIT_STRING',
    'SHORT_CIRCUIT_STRING',
    'STRING_IMBALANCE',
]
canopy_2015['label_string'] = np.select(conditions_B, choices_B, default='NORMAL')


# Check Results
print("=== Labeling B: String-Level Features ===\n")

ev = canopy_2015[evaluable]

print("String current stats (after bounds cleaning):")
print(canopy_2015[string_current_cols].describe().round(3).to_string())

print(f"\nCurrent CV distribution (daytime, POA ≥ {POA_MIN_STRING} W/m²):")
print(ev['I_string_cv'].describe().round(4))

print(f"\nRi (voltage proxy) distribution (daytime):")
print(ev[R_cols].describe().round(2).to_string())

print(f"\nR_string_mean distribution (daytime):")
print(ev['R_string_mean'].describe().round(2))

print(f"\nlabel_string distribution:")
print(canopy_2015['label_string'].value_counts())

print(f"\nPer-string mean current during daytime (POA ≥ {POA_MIN_STRING}):")
for i in range(1, N_STRINGS + 1):
    col = f'ShuntCurrent_A_Avg_{i}'
    mean_i = canopy_2015.loc[evaluable, col].mean()
    mean_r = canopy_2015.loc[evaluable, f'R_string_{i}'].mean()
    print(f"  String {i}: I = {mean_i:.2f} A   Ri = {mean_r:.1f} V")
````

## Inverter Level Features

````python
canopy_2015 = pd.read_csv('/content/drive/MyDrive/NIST_processed/canopy_2015_Label_2_corrected.csv', index_col='TIMESTAMP')
````

````python
POA_MIN_INVERTER        = 100    # W/m² — threshold for clear daylight
INV_SHUTDOWN_POWER      = 0.05   # kW — below this = inverter not producing
INV_EFFICIENCY_MIN      = 0.85   # DC→AC efficiency floor for healthy inverter
INV_EFFICIENCY_MAX      = 1.00   # ceiling — above this is physically impossible
INV_RATED_DC_KW         = 243.0  # system DC nameplate
INV_RATED_AC_KW         = 225.0  # inverter AC rated output (~0.93 × DC)

# 1. DC → AC Conversion Efficiency
# η = PAC / PDC
# Healthy inverter: 0.93–0.98 during normal operation
# Drops toward zero during startup/shutdown (normal)
# Sustained drop below 0.85 during full operation = inverter fault
# Above 1.0 is impossible
canopy_2015['inv_efficiency'] = (
    canopy_2015['InvPAC_kW_Avg'] / canopy_2015['InvPDC_kW_Avg'].replace(0, np.nan)
)
# Clip to physically meaningful range
canopy_2015['inv_efficiency_clipped'] = canopy_2015['inv_efficiency'].clip(lower=0, upper=INV_EFFICIENCY_MAX)


# 2. DC Power from Strings vs Inverter DC Input
# P_string_total = sum of all 7 shunt-measured string powers
# InvPDC_kW_Avg = inverter's own DC input measurement
# Discrepancy between the two = wiring loss, measurement error, or DC fault between the combiner box and inverter input
canopy_2015['DC_power_discrepancy'] = canopy_2015['P_string_total'] - canopy_2015['InvPDC_kW_Avg']

canopy_2015['DC_power_discrepancy_pct'] = canopy_2015['DC_power_discrepancy'] / canopy_2015['InvPDC_kW_Avg'].replace(0, np.nan) * 100


# 3. Inverter Loading Ratio
# How hard is the inverter working relative to its AC rating?
# Clipping occurs when PDC >> AC rating (common on sunny winter days)
# Persistent overloading or underloading relative to available DC = fault
canopy_2015['inv_loading_ratio'] = canopy_2015['InvPAC_kW_Avg'] / INV_RATED_AC_KW


# 4. Inverter Fault Flag Composite
# The inverter reports its own fault codes.
fault_flag_cols = [
    'InvMainFault_Max', 'InvDriveFault_Max', 'InvVoltageFault_Max',
    'InvGridFault_Max', 'InvTempFault_Max', 'InvSystemFault_Max'
]
# Any non-zero fault code = inverter self-reported fault
canopy_2015['inv_any_fault_reported'] = (canopy_2015[fault_flag_cols] > 0).any(axis=1).astype(int)
canopy_2015['inv_fault_count']        = (canopy_2015[fault_flag_cols] > 0).sum(axis=1)


# 5. Rolling Efficiency
canopy_2015['inv_efficiency_roll5'] = canopy_2015['inv_efficiency_clipped'].rolling(window=5, min_periods=3).mean()
````

````python
# 6. Inverter Fault Labels
evaluable_inv = (canopy_2015['Op_Mask'] == 1) & (canopy_2015['POA_fused'] >= POA_MIN_INVERTER)

# Inverter shutdown: daylight + irradiance present but near-zero AC output
# Most severe — system producing nothing despite available resource
inverter_shutdown = (
    evaluable_inv & (canopy_2015['InvPAC_kW_Avg'] < INV_SHUTDOWN_POWER)
)

# Efficiency fault: sustained DC→AC conversion below healthy threshold
# Excludes rows already caught by shutdown (PAC near zero makes efficiency meaningless)
inv_efficiency_fault = (
    evaluable_inv & ~inverter_shutdown
    & (canopy_2015['inv_efficiency_roll5'] < INV_EFFICIENCY_MIN)
    & (canopy_2015['InvPDC_kW_Avg'] > 1.0)
    & (canopy_2015['inv_loading_ratio'] > 0.20)
)

conditions_C = [
    canopy_2015['Op_Mask'] == 0,
    canopy_2015['InvPAC_kW_Avg'].isna() | canopy_2015['InvPDC_kW_Avg'].isna(),
    inverter_shutdown,
    inv_efficiency_fault,
]

choices_C = [
    'NOT_EVALUABLE',
    'MISSING_DATA',
    'INVERTER_SHUTDOWN',
    'INVERTER_EFFICIENCY_FAULT',
]

canopy_2015['label_inverter'] = np.select(conditions_C, choices_C, default='NORMAL')


# Check
print("=== Labeling C: Inverter-Level Features ===\n")

eff_mask = evaluable_inv & (canopy_2015['InvPDC_kW_Avg'] > 1.0)

print("DC→AC efficiency (daytime, PDC > 1 kW):")
print(canopy_2015.loc[eff_mask, 'inv_efficiency'].describe().round(4))

print(f"\nEfficiency ceiling violations (η > 1.0, daytime, PDC > 1 kW):")
above_one = (canopy_2015.loc[eff_mask, 'inv_efficiency'] > 1.0).sum()
print(f"  Count : {above_one} rows ({above_one / eff_mask.sum() * 100:.2f}% of evaluable)")

print(f"\nDC power discrepancy % (daytime):")
print(canopy_2015.loc[evaluable_inv, 'DC_power_discrepancy_pct'].describe().round(3))

print(f"\nInverter loading ratio distribution (daytime):")
print(canopy_2015.loc[evaluable_inv, 'inv_loading_ratio'].describe().round(3))

print(f"\nInverter self-reported faults:")
print(f"  Rows with any fault flag > 0 : {canopy_2015['inv_any_fault_reported'].sum()}")
print(f"  Max simultaneous fault codes : {canopy_2015['inv_fault_count'].max()}")
per_flag = (canopy_2015[fault_flag_cols] > 0).sum()
print(f"  Per-flag breakdown:\n{per_flag.to_string()}")

print(f"\nlabel_inverter distribution:")
print(canopy_2015['label_inverter'].value_counts())
print(f"\nAs % of total rows:")
print((canopy_2015['label_inverter'].value_counts() / len(canopy_2015) * 100).round(2))

print(f"\nCross-check — physics labels vs inverter self-reported faults:")
shutdown_rows = canopy_2015['label_inverter'] == 'INVERTER_SHUTDOWN'
eff_fault_rows = canopy_2015['label_inverter'] == 'INVERTER_EFFICIENCY_FAULT'
reported = canopy_2015['inv_any_fault_reported'].astype(bool)
print(f"  Shutdown detected by physics           : {shutdown_rows.sum()}")
print(f"    → also self-reported by inverter     : {(shutdown_rows & reported).sum()}")
print(f"    → NOT self-reported (silent fault?)  : {(shutdown_rows & ~reported).sum()}")
print(f"  Efficiency fault detected by physics   : {eff_fault_rows.sum()}")
print(f"    → also self-reported by inverter     : {(eff_fault_rows & reported).sum()}")
print(f"    → NOT self-reported                  : {(eff_fault_rows & ~reported).sum()}")
print(f"  Self-reported only (physics missed)    : {(~shutdown_rows & ~eff_fault_rows & reported & evaluable_inv).sum()}")
````
